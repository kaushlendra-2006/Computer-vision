```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Load dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

# Normalize images
x_train, x_test = x_train / 255.0, x_test / 255.0

# Build CNN model
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

# Compile model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train model
model.fit(x_train, y_train, epochs=30, batch_size=32, validation_split=0.2)

# Evaluate model
test_loss, test_acc = model.evaluate(x_test, y_test)
print("Test Accuracy:", test_acc)
```

What is Sharpening?

Sharpening enhances edges and fine details by increasing the contrast between neighboring pixels. It makes objects appear clearer and more defined.

A common approach is kernel-based filtering (e.g., Laplacian filter).

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
img = cv2.imread('cat.webp')
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Step 1: Blur the image (to reduce edge contrast)
blurred = cv2.GaussianBlur(img_gray, (9, 9), 0)

# Step 2: Sharpening kernel
kernel = np.array([[0, -1, 0],
                   [-1, 5, -1],
                   [0, -1, 0]])

# Step 3: Apply sharpening on blurred image
sharpened = cv2.filter2D(blurred, -1, kernel)

# Display results
plt.figure(figsize=(12,4))

plt.subplot(1, 3, 1)
plt.title("Original")
plt.imshow(img_gray, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Blurred")
plt.imshow(blurred, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Blurred + Sharpened")
plt.imshow(sharpened, cmap='gray')
plt.axis('off')

plt.show()
```

What is Contrast Adjustment?

Contrast adjustment improves visibility by stretching pixel intensity values across a wider range.

This is useful when:

The image looks washed out

Bright and dark areas are too similar

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('rose.webp', 0)

# Contrast stretching
min_val = np.min(img)
max_val = np.max(img)

contrast_stretched = (img - min_val) * (255 / (max_val - min_val))
contrast_stretched = contrast_stretched.astype(np.uint8)

# Display
plt.subplot(1, 2, 1)
plt.title("Original")
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Contrast Adjusted")
plt.imshow(contrast_stretched, cmap='gray')
plt.axis('off')

plt.show()
```

What is Histogram Equalization?

Histogram equalization redistributes pixel intensities so the histogram becomes more uniform. This enhances global contrast, especially when the image is too dark or too bright.

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('rose.webp', 0)

# Apply histogram equalization
equalized = cv2.equalizeHist(img)

# Display
plt.subplot(1, 2, 1)
plt.title("Original")
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Histogram Equalized")
plt.imshow(equalized, cmap='gray')
plt.axis('off')

plt.show()
```

Image Segmentation

Image segmentation is the process of dividing an image into meaningful regions by assigning a label to every pixel such that pixels with the same label share similar properties (intensity, color, or texture).

1. Thresholding
Explanation

Thresholding is the simplest segmentation technique.
It converts a grayscale image into a binary image based on a chosen threshold value.

Pixels above threshold → white (foreground)

Pixels below threshold → black (background)

When to use

 1. High contrast images
 2. Object clearly separable from background

```python
import cv2
import matplotlib.pyplot as plt

# Read image in grayscale
img = cv2.imread('rose.webp', 0)

# Apply thresholding
_, thresh = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)

# Display
plt.subplot(1, 2, 1)
plt.title("Original")
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Thresholded Image")
plt.imshow(thresh, cmap='gray')
plt.axis('off')

plt.show()
```

3.1. Lossy Compression
Concept

Lossy compression reduces file size by permanently removing some image information.

Some visual details are lost, but the human eye usually cannot easily notice small differences.

A very common lossy compression format is JPEG.

Key Points

Smaller file size

Some quality loss

Cannot perfectly reconstruct original image

“Create a face detector using a pre-trained Haar Cascade model for frontal faces.”

**What’s inside the XML (conceptually)**

Thousands of positive images (faces)

Thousands of negative images (non-faces)

Learned patterns like:

Eyes darker than cheeks

Nose bridge structure

Face symmetry

**Simple analogy**

CascadeClassifier = Scanner machine

.xml file = Training knowledge

**Limitations**

Works well for front-facing faces
while Struggles with:

Side faces,

Low light

```python
import cv2
# Load the Haar Cascade for face detection
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + 'haarcascade_frontalface_default.xml'
)

# Read image
img = cv2.imread("srk.jpg")

# Convert to grayscale (required for detection)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Detect faces
faces = face_cascade.detectMultiScale(
    gray,
    scaleFactor=1.3,
    minNeighbors=5
)

# Draw green rectangle around faces
for (x, y, w, h) in faces:
    cv2.rectangle(img, (x, y), (x+w, y+h), (0, 255, 0), 2)

# Show output
from google.colab.patches import cv2_imshow

cv2_imshow(img)
```

```python
# Install dependencies
!pip -q install mediapipe opencv-python  # OpenCV (image processing library) and MediaPipe (body landmark detection library )

import cv2 # Reading images, Drawing lines, Drawing circles, Writing text
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow


# Download Pose Landmarker model: Downloads pretrained Pose Landmarker model.
     # This model already knows: Human body structure, Joint locations, Pose estimation
     # No training required.

!wget -q -O pose_landmarker.task \
https://storage.googleapis.com/mediapipe-models/pose_landmarker/pose_landmarker_lite/float16/latest/pose_landmarker_lite.task


# Upload image

print("Upload a full-body image")
uploaded = files.upload()

image_path = list(uploaded.keys())[0]


# MediaPipe imports

from mediapipe.tasks import python # Loads MediaPipe Task API.
from mediapipe.tasks.python import vision # Loads vision models. Used for pose detection.
import mediapipe as mp # Main MediaPipe package.

# Load image

image = cv2.imread(image_path) # Reads uploaded image

if image is None:
    raise Exception("Unable to read image")

rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB) # OpenCV reads image in: BGR while MediaPipe expects:RGB, so conversion is necessary.

h, w, _ = image.shape  # Suppose image size: 1920 × 1080, Then: h = 1920, w = 1080

# Create MediaPipe Image: Converts image into MediaPipe format. MediaPipe models require this format.

mp_image = mp.Image(
    image_format=mp.ImageFormat.SRGB,
    data=rgb
)


# Create Pose Landmarker
# Tells MediaPipe where the relevant model file is located.

base_options = python.BaseOptions(
    model_asset_path="pose_landmarker.task"
)

# Does not create body mask. Only detect landmarks.

options = vision.PoseLandmarkerOptions(
    base_options=base_options,
    output_segmentation_masks=False
)

detector = vision.PoseLandmarker.create_from_options(options) # Creates Pose Landmarker object. Ready for prediction.


# Detect Pose

result = detector.detect(mp_image) # Main prediction step. Model analyzes image. Returns body landmarks.


# Check detection


if len(result.pose_landmarks) == 0:
    print("No human body detected")

else:

    landmarks = result.pose_landmarks[0]

    # ------------------------------------------------------
    # Helper function
    # ------------------------------------------------------

    def get_point(index):
        lm = landmarks[index]
        return np.array([
            int(lm.x * w),
            int(lm.y * h)
        ])

    # Landmark indices
    LEFT_SHOULDER = 11
    RIGHT_SHOULDER = 12

    LEFT_HIP = 23
    RIGHT_HIP = 24

    LEFT_ANKLE = 27
    RIGHT_ANKLE = 28

    NOSE = 0

    # ------------------------------------------------------
    # Extract coordinates
    # ------------------------------------------------------

    left_shoulder = get_point(LEFT_SHOULDER)
    right_shoulder = get_point(RIGHT_SHOULDER)

    left_hip = get_point(LEFT_HIP)
    right_hip = get_point(RIGHT_HIP)

    left_ankle = get_point(LEFT_ANKLE)
    right_ankle = get_point(RIGHT_ANKLE)

    nose = get_point(NOSE)

    # ------------------------------------------------------
    # Measurements
    # ------------------------------------------------------

    shoulder_width = np.linalg.norm(
        left_shoulder - right_shoulder
    )

    hip_width = np.linalg.norm(
        left_hip - right_hip
    )

    ankle_mid = (
        left_ankle + right_ankle
    ) / 2

    body_height = np.linalg.norm(
        nose - ankle_mid
    )

    shoulder_hip_ratio = shoulder_width / hip_width

    # ------------------------------------------------------
    # Simple classification
    # ------------------------------------------------------

    if shoulder_hip_ratio > 1.15:
        shape = "Athletic / Broad Shoulder"

    elif shoulder_hip_ratio < 0.90:
        shape = "Pear Shape"

    else:
        shape = "Balanced Shape"

    # ------------------------------------------------------
    # Draw landmarks
    # ------------------------------------------------------

    for lm in landmarks:

        x = int(lm.x * w)
        y = int(lm.y * h)

        cv2.circle(
            image,
            (x, y),
            4,
            (0, 255, 0),
            -1
        )

    # Shoulder line
    cv2.line(
        image,
        tuple(left_shoulder),
        tuple(right_shoulder),
        (255, 0, 0),
        3
    )

    # Hip line
    cv2.line(
        image,
        tuple(left_hip),
        tuple(right_hip),
        (0, 0, 255),
        3
    )

    # Height line
    cv2.line(
        image,
        tuple(nose),
        tuple(ankle_mid.astype(int)),
        (0, 255, 255),
        3
    )

    cv2.putText(
        image,
        shape,
        (20, 50),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (255, 255, 255),
        2
    )

    # ------------------------------------------------------
    # Results
    # ------------------------------------------------------

    print("\n========== BODY ANALYSIS ==========\n")

    print("Shoulder Width :", round(shoulder_width,2))
    print("Hip Width      :", round(hip_width,2))
    print("Body Height    :", round(body_height,2))
    print("Shoulder/Hip Ratio :", round(shoulder_hip_ratio,2))
    print("Predicted Shape :", shape)

    cv2_imshow(image)
```
